```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("incidencia_00.csv", encoding='ISO-8859-1')

filtro = df[df['indicador'] == 'Porcentaje de hogares vctimas del delito']
```

```python
numeric_cols = filtro.columns[6:-1]
filtro[numeric_cols] = filtro[numeric_cols].apply(
    lambda x:
      x.fillna(x.mean()), axis=0)

plt.figure(figsize=(15, 10))
for _, row in filtro.iterrows():
    plt.plot(numeric_cols, row[numeric_cols].values, label=row['desc_entidad'])


plt.title('Porcentaje de hogares víctimas del delito en México')
plt.xlabel('Año')
plt.ylabel('Porcentaje')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize='small', ncol=2)
plt.grid(True)
plt.tight_layout()
plt.show()
```
