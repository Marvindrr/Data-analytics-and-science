```python
active_players = batting[batting['AB'] > 0]  ## Descartamos a los jugadores sin estadisticas / inactivos del dt
active_players.describe()
```

```python
batting.groupby('playerID')['HR'].sum().sort_values(ascending=False)  ## Agrupamos para saber el tt de HR de cada jugador en su carrera ##
```

```python
babe_ruth = batting[batting['playerID'] == 'ruthba01']  ## Filtrar las estadisticas de un solo jugador ##

start_year = babe_ruth['yearID'].min()  ## Encontrar el año de inicio y final de su carrera ##
end_year = babe_ruth['yearID'].max()

ruth_years = batting[(batting['yearID'] >= start_year) & (batting['yearID'] <= end_year)]  ## Filtramos a los jugadores que jugaron en los mismos años que Ruth ##
ruth_years.groupby('playerID')['HR'].sum().sort_values(ascending=False)  ## Buscamos a los lideres de HR en esa epoca
```

```python
total_hr_by_year = batting.groupby('yearID')['HR'].sum()

plt.Figure(figsize=(10,6))
plt.plot(total_hr_by_year.index, total_hr_by_year.values)

plt.title('Total de JONRONS por año')
plt.xlabel("Año")
plt.ylabel("Total JONRONS")

plt.grid(True)
plt.show()
```

```python
## Agrupamos la informacion por equipos y año para encontrar cada HR de equipos por años
team_hr_per_year = (batting.groupby(['yearID', 'teamID']) ['HR'].sum().reindex())
```

```python
rockies_hr = team_hr_per_year.xs('LAN', level='teamID') ## Extraemos los HR de los Colorado seleccionando la seccion con .xs ##
```

```python
league_avg_hr = (
    team_hr_per_year
    .groupby('yearID')    ## Sacamos los HR de los demas equipos a partir de 1993 ##
    .mean()
)

league_avg_hr = league_avg_hr[league_avg_hr.index >= 1993]
```

```python
plt.figure(figsize=(10, 6))

plt.plot(league_avg_hr.index,
  league_avg_hr.values,
  label='AVG LIGA',           ## Creamos las estadisticas de la Liga ##
  color='orange',
  linestyle='--'
)

plt.plot(
  rockies_hr.index,
  rockies_hr.values,
  label='DADGERS',  ## Creamos las estadisticas de los dadgers##
  color="#00066F"
)

plt.title('Los DADGERS HR vs. AVG Liga')
plt.xlabel('Año')
plt.ylabel('HR')
plt.legend()

plt.grid(True)
plt.show()
```
