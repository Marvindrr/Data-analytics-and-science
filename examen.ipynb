```python
import pandas as pd
import matplotlib as plt
```

```python
df = pd.read_csv('registroMasivo.csv')
df.head()
```

```python
print(df.columns)
```

```python
FILTRO = df['Producto']. value_counts()
PRODUCTO_MAS_VENDIDO = FILTRO.idxmax()
print(PRODUCTO_MAS_VENDIDO)
```

```python
FILTRO = df['Centro Comercial']. value_counts()
CENTRO_MAS_VENTAS = FILTRO.idxmax()
CANTIDAD_VENTAS = FILTRO.max()
print(f"El centro comercial con mas ventas es: {CENTRO_MAS_VENTAS} con {CANTIDAD_VENTAS}")
```

```python
VENTAS_ESBOS = df[df['Producto'] == "xbox one"]
FILTRO = VENTAS_ESBOS['Centro Comercial']. value_counts()
FILTRO2 = FILTRO.idxmax()
print(FILTRO2)
```

```python
FILTRO = df[df['Centro Comercial'] == "benito juarez"]
MAS_VENDIDO = FILTRO['Producto']. value_counts()
FILTRO2 = MAS_VENDIDO.idxmax()
print(FILTRO2)
```
