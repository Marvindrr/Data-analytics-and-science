```python
years = [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
COLS = ['desc_entidad', 'indicador'] + [str(year) for year in years]

FILTRO = (
    (df['indicador'] == "Porcentaje de delitos con portacin de armas en personas de 18 aos y ms") &
    (df['desc_entidad'] == "Aguascalientes")
)
filtered_df = df[FILTRO][COLUMNAS]

if not filtered_df.empty:
    
    plt.figure(figsize=(10, 6))
    for index, row in filtered_df.iterrows():
        plt.plot(years, row[2:], marker='o')

    plt.xlabel('Años')
    plt.ylabel('Porcentaje')
    plt.title('Porcentaje de delitos con portación de armas en Aguascalientes')
    plt.legend()
    plt.grid(True)
    plt.show()
    print(df[FILTRO][COLUMNAS])
else:
    print("No se encuentran datos.")
```
