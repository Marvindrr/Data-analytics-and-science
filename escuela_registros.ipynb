```python
import pandas as pd
```

```python
df = pd.read_csv("becas.csv")
df.head()
```

```python
#validar alumnos con beca de 100 tengan promedio mayor a 9
#excluir alumnos con 100 pero promedio menor a 9
df_excluded_100= df[(df["promedio"]< 9) & (df["beca"] == "100%")]
df_excluded_100.index
df_new = df.copy()

df_new.loc[df_excluded_100.index,"beca"] = "0%"
df_new.loc[df_excluded_100.index,"beca"]

#Se modifica la beca a 100% de los que tienen un promedio mayor a 9 y
df_new.loc[(df["promedio"] > 9) & (df["beca"] != "100%"), "beca"] = "100%"
df_new[(df["promedio"] > 9) & (df["beca"] != "100%")]
```

```python
df_new.loc[(df["promedio"] < 9) & (df["promedio"] >8 ), "beca" ] = "75%"
df_new.loc[(df["beca"] == "75%") & (df["promedio"] < 8), "beca"] = "0%"
df_new[(df["promedio"] < 9) & (df["promedio"] >8 )]
df_new 
```

```python
df_new.loc[(df_new["promedio"] > 7) & (df_new["promedio" ] <8) & (df_new["beca"] != "50%"), "beca"] = "50%"
df_new.loc[(df_new["beca"] == "50%") & (df_new["promedio" ] <= 7) , "beca"] = "0%"
df_new
```

```python
df_new.loc[(df_new["promedio"] >= 6) & (df_new["promedio" ] <= 7) & (df_new["beca"] != "25%"), "beca"] = "25%"
df_new.loc[(df_new["beca"] == "25%") & (df_new["promedio" ] <= 6) , "beca"] = "0%"
df_new
```

```python
df_new.loc[(df_new["promedio"] < 6) , "beca"] = "0%"
df_new
```

```python
df_new.loc[(df_new["colegiatura"] == 1500) , "beca"] = "Transporte"
df_new
```

```python
#VALIDAR RESULTADOS
PROMEDIO = df_new["promedio"]
BECA = df_new["beca"]
COLEGIATURA = df_new["colegiatura"]

#No debe haber alumnos con beca del 100% y promedio menor a 9
print("Caso 1")
print(df_new[(PROMEDIO < 9) & (BECA == "100%")])
#No debe haber alumnos con beca del 75% y promedio menor a 8
print("Caso 2")
print(df_new[(PROMEDIO < 8) & (BECA == "75%")])
#No debe haber alumnos con beca del 50% y promedio menor a 7
print("Caso 3")
print(df_new[(PROMEDIO < 7) & (BECA == "50%")])
#No debe haber alumnos con beca del 25% y promedio menor a 6
print("Caso 4")
print(df_new[(PROMEDIO < 6) & (BECA == "25%")])
#No debe haber alumnos con beca si su promedio es menor a 6
print("Caso 5")
print(df_new[(PROMEDIO < 6) & (BECA != "0%")])
#No debe haber alumnos con colegiatura de 1500 y beca diferente a "transportes"
print("Caso 6")
print(df_new[(COLEGIATURA == 1500) & (BECA != "transporte")])
```

```python
df_new.to_csv("becasCleaned.csv",index=False)
```
