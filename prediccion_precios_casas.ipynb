```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('home_dataset.csv')
 
house_sizes = data ['HouseSize'].values
house_prices = data['HousePrice'].values
```

```python
plt.scatter(house_sizes, house_prices, marker ='o', color ='blue')
plt.title('House Prices vs. House Size')
plt.xlabel('House Size')
plt.ylabel('House Price')
plt.show()
```

```python
# Dividir la data para entrenamiento y tests (80-20)
x_train, x_test, y_train, y_test = train_test_split(house_sizes, house_prices, test_size=0.2, random_state=42)
```

```python
#Preparacion de datos con reshape convirtiendo el arreglo de 1D a 2D
x_train = x_train.reshape(-1, 1)
x_test = x_test.reshape(-1, 1)

#Crear el modelo predictivo 
model = LinearRegression()

#Entrenamiento del modelo predictivo 
model.fit(x_train, y_train)
```

```python
#Variable donde se guardan los resultados de la prediccion
predictions = model.predict(x_test)

plt.scatter(x_test, y_test, marker='o', color='blue', label='Precios actuales')
plt.plot(x_test, predictions, color='red', linewidth='2', label='Prediccion de precios' )
plt.title('Prediccion de precios con regresion lineal')
plt.xlabel('House Sizes')
plt.ylabel('House Prices')
plt.legend()
plt.show()
```
