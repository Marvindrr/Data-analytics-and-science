```python
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("becasCleaned.csv")
df.head()
```

```python
#Alumnos con promedio mayor a 8 con nombre apellido y beca
FILTRO = df["promedio"]>8
COLUMNAS = ["nombre", "apellido", "beca"]
df[FILTRO][COLUMNAS]
```

```python
#Alumnos con beca transporte y sacar nombre
FILTRO = df["beca"]= "transporte"
COLUMNAS = ["nombre"]
df[FILTRO][COLUMNAS]
```

```python
#Alumnos con beca 100% y sacar nombre y promedio
FILTRO = df["beca"] == "100%"
COLUMNAS = ["nombre", "promedio"]
df[FILTRO][COLUMNAS]
```

```python
#Alumnos con beca transporte y promedio mayor a 8 y sacar nombre
FILTRO = ((df["beca"] == "transporte") & df["promedio"]>8)
COLUMNAS = ["nombre"]
df[FILTRO][COLUMNAS]
```

```python
#Agrupar alumnos con diferentes tipos de becas en nuevos DataFrames
FILTRO = ["100%", "75%", "50%", "25%" "0%"]
for filtro in FILTRO:
    df_x = df[df["beca"]==filtro]
    print(df_x.head())
    df_x.to_csv(f"becas_{filtro}.csv")

FILTRO_100 = df["beca"] == "100%"
FILTRO_75 = df["beca"] == "75%"
FILTRO_50 = df["beca"] == "50%"
FILTRO_25 = df["beca"] == "25%"
FILTRO_0 = df["beca"] == "0%"
FILTRO_Transporte = df["beca"] == "transporte"

df_100 = df[FILTRO_100]
df_75 = df[FILTRO_75]
df_50 = df[FILTRO_50]
df_25 = df[FILTRO_25]
df_0 = df[FILTRO_0]
df_transporte = df[FILTRO_Transporte]

print(df_100.head())
print(df_75.head())
print(df_50.head())
print(df_25.head())
print(df_0.head())
print(df_transporte.head())
```

```python
#Graficar los alumnos en base a sus becas
becas = df["beca"].unique()
print(becas)
data = []

for beca in becas:
    FILTRO = df["beca"] == beca
    data.append([beca,df[FILTRO].count()])

df_to_graphicate = pd.DataFrame(data, columns=["BECA", "NUMERO DE ALUMNOS"])
plt.figure(figsize=(12,6))
#bar(x,y)
plt.bar(df_to_graphicate["BECA"], df_to_graphicate["NUMERO DE ALUMNOS"])
plt.xlabel("Becas")
plt.ylabel("Alumnos")
plt.title("Cantidad de alumnos por beca")
plt.show()
```
