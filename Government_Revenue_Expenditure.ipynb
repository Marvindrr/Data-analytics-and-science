```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('Gover_exp_Rev_euro.csv')

data.tail()
```

```python
#Separamos la primera columna en 5 columnas nuevas con .str.split
data[['freq', 'unit', 'sector', 'na_item', 'geo']] = data[r'freq,unit,sector,na_item,geo\TIME_PERIOD'].str.split(pat=',', expand=True)

#Reemplazar los textos nulos por valores nulos matematicos
data = data.replace(': m', np.nan)
data.head()
```

```python
# Eliminamos la primera columna que no necesitamos con .drop columns (usar .index para las filas horizontales)
data = data.drop(columns=(r'freq,unit,sector,na_item,geo\TIME_PERIOD'))
```

```python
# Creamos la lista de años para limpiar los datos de los años de manera general
lista_de_años = data.loc[:, '1995':'2025'].columns 


data.head()
```

```python
# Generalizamos los cambios con .apply
# Con .to_numeric creamos cualquier dato a numerico (con errors='coerce' le decimos que si encuentra un error no pare)
data[lista_de_años] = data[lista_de_años].apply(pd.to_numeric, errors='coerce')

data.head(5)
```
